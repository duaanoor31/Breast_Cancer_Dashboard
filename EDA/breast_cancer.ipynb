```python
print(df.info())
print(df.describe(include='all'))

df_clean = df.copy()
df_clean.columns = df_clean.columns.str.strip().str.replace(' ', '_').str.lower()

df_clean['survival_months_bin'] = pd.cut(df_clean['survival_months'], bins=[0, 30, 60, 90, 120, 150], labels=['0-30', '31-60', '61-90', '91-120', '121+'])

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
sns.histplot(data=df_clean, x='age', kde=True, ax=axes[0, 0], color='skyblue')
sns.countplot(data=df_clean, x='status', ax=axes[0, 1], palette='Set2')
sns.boxplot(data=df_clean, x='status', y='tumor_size', ax=axes[1, 0], palette='Set3')
sns.scatterplot(data=df_clean, x='age', y='tumor_size', hue='status', ax=axes[1, 1], alpha=0.7)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.violinplot(data=df_clean, x='estrogen_status', y='survival_months', hue='status', split=True, palette='muted')
plt.show()

numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(8, 6))
sns.heatmap(df_clean[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```
